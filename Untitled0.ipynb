```python
#Hacer un ejercicio en python haciendo uso de las funciones, que reste dos numeros introducidos por el usuario. TAREA.
def suma(num1, num2):
    return num1 + num2

def resta(num1, num2):
    return num1 - num2

def multiplicar(num1, num2):
    return num1 * num2

def dividir(num1, num2):
    if num2 != 0:
        return num1 / num2
    else:
        return "No se puede dividir entre cero"

def mitad(num):
    return num / 2

# Menú de operaciones
print("Selecciona una operación:")
print("1. Suma")
print("2. Resta")
print("3. Multiplicación")
print("4. División")
print("5. Mitad de un número")

opcion = input("Opción (1-5): ")

# Operaciones que requieren dos números
if opcion in ['1', '2', '3', '4']:
    num1 = float(input("Num1: "))
    num2 = float(input("Num2: "))

    if opcion == '1':
        resultado = suma(num1, num2)
    elif opcion == '2':
        resultado = resta(num1, num2)
    elif opcion == '3':
        resultado = multiplicar(num1, num2)
    elif opcion == '4':
        resultado = dividir(num1, num2)

# Operación que requiere un solo número
elif opcion == '5':
    num = float(input("Ingresa un número: "))
    resultado = mitad(num)
else:
    resultado = "Opción inválida"

print("Resultado:", resultado)
```

Tarifas de deducciones ISSS,ISR,AFP

```python
TASA_ISSS = 0.03
TASA_AFP = 0.0725
TOPE_ISSS = 30.00

TRAMOS_ISR = [
(0.01, 550.00, 0.00, 0.00, 0.00),         # Tramo 1
(550.01, 895.24, 0.10, 550.00, 17.67),    # Tramo 2
(895.25, 2038.10, 0.20, 895.24, 60.00),   # Tramo 3
(2038.11, float("inf"), 0.30, 2038.10, 288.57)  # Tramo 4
]

def calcular_isss(sueldo):
    "Calcula el descuento del ISSS (3%)"
    descuento = sueldo * TASA_ISSS
    return min(descuento, TOPE_ISSS)

def calcular_afp(sueldo):
    "Calcula el descuento del ISSS (%7.25)"
    return sueldo * TASA_AFP

def calcular_isr(sueldo):
    base = sueldo - calcular_isss(sueldo) - calcular_afp(sueldo)
    for inicio, fin, tasa, base_exenta, cuota in TRAMOS_ISR:
        if inicio <= base <= fin:
            return max(0, (base - base_exenta) * tasa + cuota)  # Aseguramos que no sea negativo
    return 0.00

print("\nCALCULADORA DE SALARIO TOTAL")
print("--------------------------")
sueldo = float(input("Ingrese sueldo Mensual: $"))

isss = calcular_isss(sueldo)
afp = calcular_afp(sueldo)
isr = calcular_isr(sueldo)
total_descuentos = isss + afp + isr
sueldo_total = sueldo - total_descuentos

print("\nDESGLOSE DE SALARIO")
print("+------------------+-----------+")
print("| Concepto         | Valor     |")
print("+------------------+-----------+")
print(f"| Sueldo Mensual   | ${sueldo:>9.2f} |")
print(f"| ISSS (3%)        | ${isss:>9.2f} |")
print(f"| AFP (7.25%)      | ${afp:>9.2f} |")
print(f"| ISR              | ${isr:>9.2f} |")
print("+------------------+-----------+")
print(f"| Total descuentos | ${total_descuentos:>9.2f} |")
print(f"| Sueldo total     | ${sueldo_total:>9.2f} |")
print("+------------------+-----------+")







```

```python
# Diccionarios base
tasas_conversion = {
    "USD": 1.0, "EUR": 0.85, "GBP": 0.74, "CHF": 0.91,
    "JPY": 110.27, "HKD": 7.77, "CAD": 1.25, "CNY": 6.47, "COL": 8.75, "MXN": 16.9
}

longitud = {"metro": 1, "cm": 0.01, "mm": 0.001, "km": 1000, "pulgada": 0.0254, "pie": 0.3048, "yarda": 0.9144, "milla": 1609.34, "nanometro": 1e-9, "micrometro": 1e-6}
masa = {"kg": 1, "g": 0.001, "mg": 1e-6, "lb": 0.4536, "oz": 0.0283, "tonelada": 1000, "quintal": 45.36, "stone": 6.35, "microgramo": 1e-9}
almacenamiento = {"byte": 1, "kb": 1024, "mb": 1024**2, "gb": 1024**3, "tb": 1024**4, "bit": 0.125, "nibble": 0.5, "pb": 1024**5, "eb": 1024**6}
tiempo = {"segundo": 1, "minuto": 60, "hora": 3600, "día": 86400, "semana": 604800, "mes": 2.63e6, "año": 3.15e7, "milisegundo": 0.001, "microsegundo": 1e-6}
volumen = {"litro": 1, "ml": 0.001, "galón": 3.785, "onza": 0.0296, "m3": 1000, "cm3": 0.001, "taza": 0.24, "pinta": 0.473, "cuarto": 0.946}
area = {"m2": 1, "cm2": 0.0001, "mm2": 1e-6, "km2": 1e6, "hectarea": 10000, "acre": 4046.86, "pie2": 0.0929, "pulgada2": 0.000645, "yarda2": 0.836, "manzana": 6987.4}
manzanas = {"manzana": 1, "m2": 1/6987.4, "hectarea": 1/0.6987, "acre": 1/1.726, "pie2": 1/75264, "km2": 1/143, "cm2": 1/69874000, "mm2": 1/698740000, "tarea": 1/6.98}
tareas = {"tarea": 1, "m2": 1/350, "manzana": 1/6.98, "hectarea": 1/3.5, "acre": 1/1.4, "pie2": 1/37674, "km2": 1/286, "cm2": 1/3500000, "mm2": 1/35000000}

# Configuración para cálculo de salario
TASA_ISSS = 0.03
TASA_AFP = 0.0725
TOPE_ISSS = 30.00

TRAMOS_ISR = [
    (0.01, 550.00, 0.00, 0.00, 0.00),         # Tramo 1
    (550.01, 895.24, 0.10, 550.00, 17.67),    # Tramo 2
    (895.25, 2038.10, 0.20, 895.24, 60.00),   # Tramo 3
    (2038.11, float("inf"), 0.30, 2038.10, 288.57)  # Tramo 4
]

# Funciones de conversión
def convertir_moneda(cantidad, origen, destino):
    if origen not in tasas_conversion or destino not in tasas_conversion:
        return None
    usd = cantidad / tasas_conversion[origen]
    return usd * tasas_conversion[destino]

def convertir_general(cantidad, origen, destino, diccionario):
    if origen not in diccionario or destino not in diccionario:
        return None
    base = cantidad * diccionario[origen]
    return base / diccionario[destino]

def conversor_personalizado(tipo, cantidad):
    if tipo == "caja":
        return f"{int(cantidad)} unidades = {int(cantidad) // 25} cajas y {int(cantidad) % 25} unidades"
    elif tipo == "saco":
        return f"{int(cantidad)} libras = {int(cantidad) // 50} sacos y {int(cantidad) % 50} libras"
    elif tipo == "fardo":
        return f"{int(cantidad)} unidades = {int(cantidad) // 5} fardos y {int(cantidad) % 5} unidades"
    else:
        return "Tipo no válido."

def calcular_agua(metros):
    if metros <= 18:
        return 6.00
    elif metros <= 28:
        return 6.00 + (metros - 18) * 0.45
    else:
        return 6.00 + 10 * 0.45 + (metros - 28) * 0.65

def tabla_meses(n):
    print("\nMes\tAcumulado")
    for mes in range(1, n+1):
        acumulado = mes * (mes + 1) // 2
        print(f"{mes}\t{acumulado}")

# Funciones para cálculo de salario
def calcular_isss(sueldo):
    descuento = sueldo * TASA_ISSS
    return min(descuento, TOPE_ISSS)

def calcular_afp(sueldo):
    return sueldo * TASA_AFP

def calcular_isr(sueldo):
    base = sueldo - calcular_isss(sueldo) - calcular_afp(sueldo)
    for inicio, fin, tasa, base_exenta, cuota in TRAMOS_ISR:
        if inicio <= base <= fin:
            return max(0, (base - base_exenta) * tasa + cuota)
    return 0.00

def calcular_salario():
    print("\nCALCULADORA DE SALARIO POR DEDUCCIONES")
    print("--------------------------")
    try:
        sueldo = float(input("Ingrese sueldo Mensual: $"))

        isss = calcular_isss(sueldo)
        afp = calcular_afp(sueldo)
        isr = calcular_isr(sueldo)
        total_descuentos = isss + afp + isr
        sueldo_total = sueldo - total_descuentos

        print("\nDESGLOSE DE SALARIO")
        print("+------------------+-----------+")
        print("| Concepto         | Valor     |")
        print("+------------------+-----------+")
        print(f"| Sueldo Mensual   | ${sueldo:>9.2f} |")
        print(f"| ISSS (3%)        | ${isss:>9.2f} |")
        print(f"| AFP (7.25%)      | ${afp:>9.2f} |")
        print(f"| ISR              | ${isr:>9.2f} |")
        print("+------------------+-----------+")
        print(f"| Total descuentos | ${total_descuentos:>9.2f} |")
        print(f"| Sueldo total     | ${sueldo_total:>9.2f} |")
        print("+------------------+-----------+")
    except ValueError:
        print("Error: Ingrese un valor numérico válido")

# Menú principal
def menu():
    print("\n--- Menú de opciones ---")
    print("1. Conversor de monedas")
    print("2. Conversores físicos")
    print("3. Conversores personalizados")
    print("4. Consumo de agua potable")
    print("5. Tabla de meses acumulados")
    print("6. Calculadora de salario por deducciones")
    print("7. Salir")
    return input("Seleccione una opción (1-7): ")

# Ejecución principal
while True:
    opcion = menu()

    if opcion == "1":
        print("Monedas disponibles:", ", ".join(tasas_conversion.keys()))
        o = input("Moneda origen: ").upper()
        if o not in tasas_conversion:
            print("⚠️ Moneda de origen no disponible.")
        else:
            d = input("Moneda destino: ").upper()
            if d not in tasas_conversion:
                print("⚠️ Moneda de destino no disponible.")
            else:
                try:
                    c = float(input("Cantidad: "))
                    r = convertir_moneda(c, o, d)
                    print(f"{c} {o} = {r:.2f} {d}")
                except ValueError:
                    print("⚠️ Entrada inválida. Solo números.")

    elif opcion == "2":
        grupos = [longitud, masa, almacenamiento, tiempo, volumen, area, manzanas, tareas]
        nombres = ["Longitud", "Masa", "Almacenamiento", "Tiempo", "Volumen", "Área", "Manzanas", "Tareas"]
        for i, nombre in enumerate(nombres, 1):
            print(f"{i}. {nombre}")
        try:
            sub = int(input("Seleccione tipo (1-8): "))
            dic = grupos[sub - 1]
            print("Unidades disponibles:", ", ".join(dic.keys()))
            o = input("Unidad origen: ").lower()
            d = input("Unidad destino: ").lower()
            if o not in dic:
                print("⚠️ Unidad de origen no disponible.")
            elif d not in dic:
                print("⚠️ Unidad de destino no disponible.")
            else:
                c = float(input("Cantidad: "))
                r = convertir_general(c, o, d, dic)
                print(f"{c} {o} = {r:.4f} {d}")
        except (IndexError, ValueError):
            print("⚠️ Entrada inválida.")

    elif opcion == "3":
        tipo = input("Tipo (caja, saco, fardo): ").lower()
        if tipo not in ["caja", "saco", "fardo"]:
            print("⚠️ Tipo no válido.")
        else:
            try:
                cantidad = int(input("Cantidad: "))
                print(conversor_personalizado(tipo, cantidad))
            except ValueError:
                print("⚠️ Entrada inválida. Solo enteros.")

    elif opcion == "4":
        try:
            metros = float(input("Metros cúbicos consumidos: "))
            print(f"Costo total: ${calcular_agua(metros):.2f}")
        except ValueError:
            print("⚠️ Entrada inválida. Solo números.")

    elif opcion == "5":
        try:
            n = int(input("¿Cuántos meses desea acumular?: "))
            if n < 1:
                print("⚠️ Ingrese un número positivo.")
            else:
                tabla_meses(n)
        except ValueError:
            print("⚠️ Entrada inválida. Solo enteros.")

    elif opcion == "6":
        calcular_salario()

    elif opcion == "7":
        print("Gracias por usar el sistema. ¡Hasta pronto!")
        break

    else:
        print("⚠️ Opción inválida. Intente nuevamente.")

    if input("\n¿Desea realizar otra operación? (s/n): ").lower() != "s":
        print("Gracias por usar el sistema. ¡Hasta pronto!")
        break
```

```python
def calcular_varianza():
    """
    Calcula la varianza de una serie de números introducidos por el usuario.
    """
    numeros = []

    # Solicitar al usuario que ingrese los números
    print("Ingresa los números para calcular la varianza. Escribe 'fin' para terminar.")
    while True:
        entrada = input("Introduce un número: ")
        if entrada.lower() == 'fin':
            break
        try:
            numeros.append(float(entrada))
        except ValueError:
            print("Entrada inválida. Por favor, ingresa un número válido o 'fin'.")

    if not numeros:
        print("No se introdujeron números. No se puede calcular la varianza.")
        return

    # Calcular la media
    media = sum(numeros) / len(numeros)

    # Calcular las diferencias al cuadrado
    diferencias_cuadrado = [(x - media) ** 2 for x in numeros]

    # Calcular la varianza
    varianza = sum(diferencias_cuadrado) / len(numeros)

    print(f"\nLos números ingresados son: {numeros}")
    print(f"La media (promedio) es: {media:.2f}")
    print(f"La varianza es: {varianza:.2f}")

# Ejecutar la función
calcular_varianza()
```

```python
def encontrar_frecuencia():
    """
    Encuentra la frecuencia de una serie de números introducidos por el usuario.
    """
    numeros = []

    # Solicitar al usuario que ingrese los números
    print("Ingresa los números para encontrar la frecuencia. Escribe 'fin' para terminar.")
    while True:
        entrada = input("Introduce un número: ")
        if entrada.lower() == 'fin':
            break
        try:
            numeros.append(int(entrada))
        except ValueError:
            print("Entrada inválida. Por favor, ingresa un número entero o 'fin'.")

    if not numeros:
        print("No se introdujeron números. No se puede calcular la frecuencia.")
        return

    frecuencia = {}
    for numero in numeros:
        frecuencia[numero] = frecuencia.get(numero, 0) + 1

    print(f"\nLos números ingresados son: {numeros}")
    print("\nFrecuencia de los números:")
    for num, freq in frecuencia.items():
        print(f"El número {num} aparece {freq} vez(ces).")

# Ejecutar la función
encontrar_frecuencia()
```

```python
def es_palindromo():
    """
    Determina si una palabra introducida por el usuario es un palíndromo.
    """
    palabra = input("Ingresa una palabra o frase para verificar si es un palíndromo: ")

    # Limpiar la palabra (quitar espacios y convertir a minúsculas)
    palabra_limpia = palabra.replace(" ", "").lower()

    # Invertir la palabra
    palabra_invertida = palabra_limpia[::-1]

    # Comparar la palabra limpia con su versión invertida
    if palabra_limpia == palabra_invertida:
        print(f"¡'{palabra}' es un palíndromo! 🥳")
    else:
        print(f"'{palabra}' no es un palíndromo. 😔")

# Ejecutar la función
es_palindromo()
```

```python
#Calculo de impuesto a las actividades economicas

TARIFAS = [
    # (desde, hasta, precio, adicional)
    (0.01, 500.00, 1.5, 0),
    (500.01, 1000.00, 1.5, 3),
    (1000.01, 2000.00, 3, 3),
    (2000.01, 3000.00, 6, 3),
    (3000.01, 6000.00, 9, 2),
    (6000.01, 8000.00, 15, 2),
    (8000.01, 18000.00, 15, 2),
    (18000.01, 30000.00, 39, 2),
    (30000.01, 60000.00, 63, 1),
    (60000.01, 100000.00, 93, 0.8),
    (100000.01, 200000.00, 125, 0.7),
    (200000.01, 300000.00, 195, 0.6),
    (300000.01, 400000.00, 255, 0.45),
    (400000.01, 500000.00, 300, 0.4),
    (500000.01, 1000000.00, 340, 0.3),
    (1000000.01, 99999999.00, 490, 0.18)
]

def calcular_tarifa(valor):
    """
    Calcula la tarifa basada en el valor ingresado según la tabla de escalones
    Fórmula: Precio + [(Valor - Desde) / 1000] * Adicional
    """
    for desde, hasta, precio, adicional in TARIFAS:
        if desde <= valor <= hasta:
            excedente = valor - desde
            monto_adicional = (excedente / 1000) * adicional
            return round(precio + monto_adicional, 2)

    return 0.00

def main():
    print("\nCALCULADORA DE TARIFAS")
    print("-----------------------------------")

    try:
        valor = float(input("Ingrese el valor a calcular: $"))

        if valor <= 0:
            print("Error: El valor debe ser mayor a 0")
            return

        tarifa = calcular_tarifa(valor)

        # Encontrar el rango aplicado para mostrar detalles
        for desde, hasta, precio, adicional in TARIFAS:
            if desde <= valor <= hasta:
                excedente = valor - desde
                monto_adicional = (excedente / 1000) * adicional
                break

        print("\nDESGLOSE DEL CÁLCULO")
        print("+----------------------+-----------+")
        print("| Concepto             | Valor     |")
        print("+----------------------+-----------+")
        print(f"| Valor base           | ${valor:>9.2f} |")
        print(f"| Rango aplicado       | ${desde:>9.2f} - ${hasta:>9.2f} |")
        print(f"| Precio fijo          | ${precio:>9.2f} |")
        print(f"| Excedente            | ${excedente:>9.2f} |")
        print(f"| Tasa adicional       | ${adicional:>9.2f} por 1000 |")
        print(f"| Monto adicional      | ${monto_adicional:>9.2f} |")
        print("+----------------------+-----------+")
        print(f"| TOTAL A PAGAR        | ${tarifa:>9.2f} |")
        print("+----------------------+-----------+")

    except ValueError:
        print("Error: Por favor ingrese un valor numérico válido")

if __name__ == "__main__":
    main()


```

Conversor De Area


```python
UNIDADES = {
    "piecuadrado": 0.092903,
    "varacuadrada": 0.6987,
    "yardacuadrada": 0.836127,
    "metrocuadrado": 1.0,
    "tarea": 698.74,
    "manzana": 6987.4,
    "hectarea": 10000.0
}

opciones = {
    "1": "piecuadrado",
    "2": "varacuadrada",
    "3": "yardacuadrada",
    "4": "metrocuadrado",
    "5": "tarea",
    "6": "manzana",
    "7": "hectarea"
}

nombres = {
    "piecuadrado": "Pies Cuadrados",
    "varacuadrada": "Varas Cuadradas",
    "yardacuadrada": "Yardas Cuadradas",
    "metrocuadrado": "Metros Cuadrados",
    "tarea": "Tareas",
    "manzana": "Manzanas",
    "hectarea": "Hectáreas"
}

print("CONVERSOR DE UNIDADES DE ÁREA")
print("==========================================")

print("\nUNIDADES DISPONIBLES:")
print("--------------------")
for i in range(1, 8):
    print(f"{i}. {nombres[opciones[str(i)]]}")
print("--------------------")

valor = float(input("\nIngrese el valor a convertir: "))

opcion_origen = input("\nSeleccione la unidad de ORIGEN (1-7): ")
opcion_destino = input("Seleccione la unidad de DESTINO (1-7): ")

# Convertir directamente sin validar
unidad_origen = opciones[opcion_origen]
unidad_destino = opciones[opcion_destino]

# Calcular resultado
resultado = (valor * UNIDADES[unidad_origen]) / UNIDADES[unidad_destino]

print(f"\nRESULTADO:")
print(f"{valor} {nombres[unidad_origen]} = {resultado:.6f} {nombres[unidad_destino]}")
```

Cajero automatico

```python
def simular_cajero():
    try:
        cantidad = float(input("Ingrese la cantidad a retirar: "))
    except ValueError:
        return

    if cantidad <= 0:
        return

    print(f"Desglose para ${cantidad:.2f}:")

    denominaciones_billetes = [20, 10, 5, 1]
    denominaciones_monedas = [0.25, 0.10, 0.05, 0.01]

    cantidad_restante = cantidad

    # Billetes
    for billete in denominaciones_billetes:
        if cantidad_restante >= billete:
            cantidad_billetes = int(cantidad_restante // billete)
            print(f"{cantidad_billetes} billete(s) de a ${billete}")
            cantidad_restante %= billete

    # Coins
    cantidad_restante_centavos = round(cantidad_restante * 100)

    for moneda_centavos in [int(m * 100) for m in denominaciones_monedas]:
        if cantidad_restante_centavos >= moneda_centavos:
            cantidad_monedas = int(cantidad_restante_centavos // moneda_centavos)
            valor_moneda_dolar = moneda_centavos / 100
            print(f"{cantidad_monedas} moneda(s) de a ${valor_moneda_dolar:.2f} ctvs.")
            cantidad_restante_centavos %= moneda_centavos

simular_cajero()
```
